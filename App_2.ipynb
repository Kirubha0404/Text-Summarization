from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer



text = """
Artificial intelligence is one of the most important technologies in the modern world.
It is used in many areas such as healthcare, education, banking, transportation, and entertainment.
AI systems can analyze large amounts of data and make useful predictions.
In healthcare, artificial intelligence can help doctors identify diseases and analyze medical images.
In education, AI can provide personalized learning experiences for students.
Companies also use AI chatbots to communicate with customers and provide support.
As technology continues to develop, artificial intelligence is expected to become an important part of everyday life.
"""


# Create a parser
parser = PlaintextParser.from_string(text, Tokenizer("english"))

# Create summarizer
summarizer = LsaSummarizer()

# Generate summary
summary = summarizer(parser.document, 3)

print("Original Text:\n")
print(text)

print("\nSummary:\n")

for sentence in summary:
    print(sentence)
